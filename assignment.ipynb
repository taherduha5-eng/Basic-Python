```python
# 1. Write a function that takes a list of strings and a number K. The function should return a new list containing only the strings that have a length greater than K.
# • Input: list=["apple", "pie", "banana"], K=3
# • Output: [apple,banana ]
```

```python
class StringFilter:
    def __init__(self, words, k):
        self.words = words
        self.k = k

    def filter_strings(self):
        result = []
        for word in self.words:
            if len(word) > self.k:
                result.append(word)
        return result
```

```python
obj = StringFilter(["apple", "pie", "banana"], 3)
print(obj.filter_strings())
```

```python
# 2. Write a function named area that calculates the area of either a "circle" or a
# "rectangle". The function should take the shape name as the first argument, followed by the necessary dimensions (radius for circle; length and width for rectangle).
# Example Input: area"circle", 5) OR area"rectangle", 4, 6)
# Example Output: Circle Area: 78.5, Rectangle Area: 24
```

```python
import math

class Shape:
    def __init__(self, shape_type, *dimensions):
        self.shape_type = shape_type
        self.dimensions = dimensions

    def area(self):
        if self.shape_type == "circle":
            radius = self.dimensions[0]
            return math.pi * radius * radius

        elif self.shape_type == "rectangle":
            length = self.dimensions[0]
            width = self.dimensions[1]
            return length * width
    def __init__(self, shape_type, *dimensions):
        self.shape_type = shape_type
        self.dimensions = dimensions

    def area(self):
        if self.shape_type == "circle":
            radius = self.dimensions[0]
            return math.pi * radius * radius

        elif self.shape_type == "rectangle":
            length = self.dimensions[0]
            width = self.dimensions[1]
            return length * width
```

```python

    
circle = Shape("circle", 5)
print(f"Circle Area: {circle.area():.1f}")   

rectangle = Shape("rectangle", 4, 6)
print(f"Rectangle Area: {rectangle.area()}")
```

```python
# Write a program that takes an email address as input and splits it into two parts: the
# "username" and the "domain".
# • Example Input: support@google.com
# • Example Output: Username: support, Domain: google.com
```

```python
class EmailSplitter:
    def __init__(self, email):
        self.email = email

    def split_email(self):
        parts = self.email.split("@")
        username = parts[0]
        domain = parts[1]
        return username, domain
```

```python
email_input = "support@google.com"
obj = EmailSplitter(email_input)
username, domain = obj.split_email()

print("Username:", username)
print("Domain:", domain)
```

```python
# 4. Write a program to find the sum of all numbers in a list. ensure the program ignores any elements in the list that are not numbers (like strings or booleans).
# • Example Input: [10, "apple", 20, True, 5]
# Example Output: Total Sum: 35
```

```python
my_list = [10, "apple", 20, True, 5]

total = 0
for item in my_list:
    if type(item) == int:  
        total += item

print("Total Sum:", total)
```

```python
# Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.)
# Perform the following tasks now:
# Now add items to the menu.
# Make table reservations.
# Take customer orders.
# Print the menu.
```

```python
class Restaurant:
    def __init__(self, name):
        self.name = name
        self.menu_items = {}      
        self.book_table = []      
        self.customer_orders = [] 

    
    def add_item_to_menu(self, item_name, price):
        self.menu_items[item_name] = price
        print(f"Added {item_name} to menu with price {price}.")

    def book_tables(self, customer_name, table_number):
        self.book_table.append({'customer': customer_name, 'table': table_number})
        print(f"Table {table_number} booked for {customer_name}.")

    def customer_order(self, customer_name, order_items):
    
        self.customer_orders.append({'customer': customer_name, 'order': order_items})
        print(f"{customer_name} ordered: {order_items}")

    def print_menu(self):
        print(f"\nMenu of {self.name}:")
        for item, price in self.menu_items.items():
            print(f"{item} - ${price}")
```

```python

my_restaurant = Restaurant("Tasty Bites")

my_restaurant.add_item_to_menu("Burger", 10)
my_restaurant.add_item_to_menu("Pizza", 15)
my_restaurant.add_item_to_menu("Pasta", 12)

my_restaurant.book_tables("Alice", 1)
my_restaurant.book_tables("Bob", 2)

my_restaurant.customer_order("Alice", ["Burger", "Pasta"])
my_restaurant.customer_order("Bob", ["Pizza"])

my_restaurant.print_menu()
```

```python
# A Bike Rental System Let us start writing the class Customer that is composed of the following attributes:
# • name
# • age
# Here are some of the methods that you need to add to your class Customer;
# • def .
# _init_(self).
# • Getter / setter for all attributes.
# class MainBikeRental that is composed of the following attributes:
# List of bikes
# List of Rentals
# Here are some of the methods that you need to add to your class MainBikeRental;
# • def _init_(self.
# • Getter / setter for all attributes.
# requestBike: Takes a request from the customer for the number of bikes and check if the requested number is available or not if yes, function will deduct number_of_bikes from shop stock else it will print wait 10 mins. And you have to make sure that the age of the customer is over than 6 years.
# returnBike: Allows customers to return their bikes to the rental shop by taking rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying rentalTime * 40. Also, add the returned bikes to the stock store again addBike: adds bikes into the list
# • totalCost: returns the total cost of bikes in the store class Bike that is composed of the following attributes:
# • name price
# Here are some of the methods that you need to add to your class Bike;
# • def _init_(self).
# Getter / setter for all attributes.
```

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    
    def get_name(self):
        return self.name
    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age
    def set_age(self, age):
        self.age = age

# =================== Class Bike 
class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    
    def get_name(self):
        return self.name
    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price
    def set_price(self, price):
        self.price = price

# =================== Class MainBikeRental 
class MainBikeRental:
    def __init__(self):
        self.list_of_bikes = []    
        self.list_of_rentals = []  

    
    def get_bikes(self):
        return self.list_of_bikes
    def set_bikes(self, bikes):
        self.list_of_bikes = bikes

    def get_rentals(self):
        return self.list_of_rentals
    def set_rentals(self, rentals):
        self.list_of_rentals = rentals

    
    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() < 6:
            print(f"{customer.get_name()} is too young to rent a bike.")
            return
        if num_of_bikes <= len(self.list_of_bikes):
            rented_bikes = self.list_of_bikes[:num_of_bikes]
            self.list_of_bikes = self.list_of_bikes[num_of_bikes:]
            self.list_of_rentals.append({'customer': customer.get_name(), 'bikes': rented_bikes})
            print(f"{customer.get_name()} rented {num_of_bikes} bike(s).")
        else:
            print("Not enough bikes available. Wait 10 minutes.")

    
    def returnBike(self, rentalTime, num_of_bikes, customer_name):
        self.list_of_bikes += [Bike(f"Bike{i+1}", 40) for i in range(num_of_bikes)]
        cost = rentalTime * 40 * num_of_bikes
        print(f"{customer_name} returned {num_of_bikes} bike(s). Total cost: ${cost}")

    def addBike(self, bike):
        self.list_of_bikes.append(bike)
    def totalCost(self):
        total = sum(bike.get_price() for bike in self.list_of_bikes)
```

```python
c1 = Customer("Alice", 10)
c2 = Customer("Bob", 5)  

b1 = Bike("Bike1", 40)
b2 = Bike("Bike2", 40)
b3 = Bike("Bike3", 40)

 
rental_shop = MainBikeRental()

rental_shop.addBike(b1)
rental_shop.addBike(b2)
rental_shop.addBike(b3)

rental_shop.requestBike(c1, 2)  
rental_shop.requestBike(c2, 1) 
rental_shop.returnBike(rentalTime=2, num_of_bikes=2, customer_name="Alice")

print("Total cost of bikes in store:", rental_shop.totalCost())
```
