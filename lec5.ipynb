```python
#  # 1. create person class with attributes(name , age, gender ) , methods welcome MISS / MR / other based on gender 
#  , and print_all_info method
```

```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    
    def welcome(self):
        if self.gender =="male":
            return f"Welcome MR{self.name}"
        elif self.gender == "female":
            return f"Welcome MISS{self.name}"
        else:
            return f"Welcome{self.name}"
    
    def print_all_info(self):
        print(f"My  name  is {self.name},  My age  is  { self.age }, My  gender  is  {self.gender}")   
```

```python
P = Person("Duha", 20 , "female")
```

```python
P.welcome()
```

```python
P.print_all_info()
```

```python
# 2. EX :

# Create a class named Car.

# Each car should have these attributes: brand, model, and year.

# Add a method named start_engine() that prints:
# "The engine of [brand] [model] is now running!"

# Add another method called car_info() that prints:
# "This car is a [year] [brand] [model]."

# Create at least 2 objects from this class and call their methods.
```

```python
class Car:
    def __init__(self,brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
    
    def start_engine(self):
        print(f"The engine of {self.brand} {self.model} is now running!")
    
    def car_info(self):
        print(f" This car is a {self.year}, {self.brand}, {self.model}.")
    
    
```

```python
Car1 = Car("mg","hs",2023)
Car1.start_engine()
```

```python
Car1.car_info()
```

```python
#  + ,-,/,*


class  operation :
    @staticmethod
    def  ADD (x,  y ) :  
        print(x +y )
    @staticmethod
    def  SUB (x,  y ) :
        print(x-y)
```

```python
# task 1: 
#create a class student , has attributes  name age and grades[math , arabic , engligh ] , 
# and has methodes display the information and another one to get the total grade from 100%.
# I need to get also the total num of students 
```

```python
class student:
    total_students = 0
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades
        student.total_students+=1
    
    def display_info(self) :
        print(f"My name is {self.name}, My age is {self.age}, My grades are {self.grades}")
        
    def total_grade(self):
        total = len(self.grades)*100 
        return  f"the total grade from 100%  is   { (sum(self.grades)/total)*100}"
              
        
```

```python

st  =  student("duha" , 20 ,  [50, 90, 60]) 
```

```python
st.display_info()
```

```python
st.total_grade()
```
